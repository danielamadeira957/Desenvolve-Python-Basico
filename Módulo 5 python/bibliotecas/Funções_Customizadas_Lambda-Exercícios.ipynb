### Q2.

- Escreva uma função em Python chamada ```soma_quadrados``` que recebe dois números como parâmetros e retorna a soma dos seus quadrados. 
- No programa principal solicite ao usuário que insira dois números e utilize a função para exibir a soma dos quadrados.

````python
## Crie aqui a função soma_quadrados()
def soma_quadrados (num1, num2):
    #calcula o quadrado de cada número e retorna a soma
    resultado = (num1 ** 2) + (num2 ** 2)
    return resultado
````

````python
## Escreva aqui o programa principal e execute a célula
#solicita a entrada do usuário
n1 = float (input("Digite o primeiro número:"))
n2 = float (input("Digite o segundo número"))

#chama a função e armazena o resultado
resultado_final = soma_quadrados (n1, n2)

#exibe o resultado na tela
print (f"A soma dos quadrados de {n1} e {n2} é: {resultado_final}")
````

### Q3.

- Crie uma função em Python chamada ```soma_digitos``` que recebe um número inteiro como parâmetro e retorna a soma dos seus dígitos. Por exemplo, para o número 123, a função deve retornar 6, $(1 + 2 + 3)$.
    - O desafio aqui é separar os dígitos de um número inteiro usando operações aritméticas <br> <br>

- No programa principal solicite ao usuário que insira um número e utilize a função ```soma_digitos``` para calcular e exibir a soma dos seus dígitos.

````python
def soma_digitos (numero):
    soma = 0
    #garantir que o número seja positivo para o cálculo
    numero = abs (numero)

    while numero > 0:
        #pega o último dígito (resto da divisão por 10)
        digito = numero % 10
        soma += digito
        #remove o último dígito (divisão inteira por 10)
        numero = numero // 10

        return soma
````

````python
#solicita a entrada do úsuario
num = int(input("Digite um número inteiro:"))
#calcula e exibe o resultado
resultado = soma_digitos(num)
print(f"Asoma dos dígitos de {num} é {resultado}")
````

### Q6.

Crie uma função lambda para verificar se um número é par ou ímpar. Em seguida, solicite ao usuário um número indefinidos de valores (até que o usuário digite 0). Para cada valor de entrada, informe se é par ou ímpar.

Exemplo de interação:
```
Digite os valores que deseja verificar a paridade (digite 0 para finalizar a entrada de dados):
3
ímpar
8
par
12
par
5
ímpar
7ímpar
```

````python
#criação da função lambda para verificar paridade
#retorna "par" se o resto da divisão por 2 for 0,  caso contário retorna "impar"
verificar_paridade = lambda x :"par" if x % 2 == 0 else "impar"

print ("Digite os valores que deseja verificar a paridade (digite 0 para finalizar a entrada de dados):")

#loop para solicitar números indefinidamente
while True:
    try:
        num=int (input())

        #condição de parada:se o usuário digitar o 
        if num == 0:
            break

        #informa se o valor de entrada é par ou ímpar usando a função lambda
        resultado = verificar_paridade(num)
        print(resultado)

    except ValueError:
        print("Por favor, insira um número inteiro válido.")
````

Escreva um programa que pergunte ao usuário qual operação ele deseja: maior ou menor.
Em seguida leia uma quantidade indefinida de valores do usuário, até que o usuário digite o valor zero.
Apresente ao final o maior ou menor dos valores digitados de acordo com a escolha do usuário.

**Sua solução deve incluir pelo menos uma função ```lambda```**

```

Digite os valores de entrada. Digite 0 para finalizar a entrada de valores.
4
12
15
1
0

O maior valor é: 15
```

````python
#define as operações usando lambda
operacoes = {
    "1": lambda lista: max (lista),
    "2": lambda lista: min (lista)
}

#pergunta a operação ao usuário
print("Opções: (1) maior ou (2)menor?")
opcao = input ("Opção:")

if opcao in operacoes:
    valores = []
    print("Digite os valores de entrada. Digite 0 para finalizar.")

    while True:
        num = int (input())
        if num == 0:
            break
        valores.append(num)

        if valores:
            resultado = operacoes [opcao](valores)
            rotulo = "maior" if opcao == "1" else "menor"
            print (f"O {rotulo} valor é:{resultado}")
        else:
            print("Nenhum valor foi digitado.")
else:
    print ("Opção inválida.")
````
